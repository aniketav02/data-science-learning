- Exercise 2: Positive or Negative

- Ask the user for a number.

- If number is positive → print "Positive"
- Else → print "Negative or Zero"

```python
num = int(input("Enter number:"))

if num >  0:
    print("Positive number")
elif num < 0:
    print("Negative number")
else:
    print("Zero")
```
